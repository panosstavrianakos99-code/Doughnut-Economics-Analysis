# Social and Ecological Indicators: A Doughnut Economics View

This notebook explores global social shortfalls and ecological pressures using indicator data spanning roughly 2000–2022. It moves from global summaries to country-group comparisons, then presents the findings through a sandwich chart and an interactive doughnut view.

## Reading guide

- Global social indicators and ecological boundaries- Ecological overshoot by country-income group- Social shortfalls by country-income group- Global sandwich comparison- 2022 doughnut view and indicator units

## Scope and interpretation

Indicators have different units, coverage periods, and definitions. Treat the charts as descriptive comparisons, not causal evidence. Some series begin after 2000; interpret each indicator using its own coverage and unit.

# Data Analysis and Visualization of Ecological and Social Indicators between 2000 and 2022.

Society faces several obstacles according to it self's factors. On the pivot table below we see the percentage of the world's population of having or not interaction with these factors.

To create this table, firstly I read the relevant csv file and in order to manage less columns I dropped the redundant ones. After that, I found the min and max of the dates that have values and I merged them again with the initial dataframe. In order to create the final pivot table I had to rename the indicators, sort the values and use as an aggregation function the min and max of the values. Lastly, I configurated the visual parameters.

```python
from IPython.display import display,HTML
import pandas as pd
from pandas import pivot_table
df = pd.read_csv('3_20250108_doughnutData.csv')
table_1 = df[(df['domain'] == 'social') & (df['type'] == 'global doughnut')].reset_index(drop=True)
table_1['dimension']= table_1['dimension'].astype('string')
table_1['indicator']= table_1['indicator'].astype('string')
table_1 = table_1.drop(columns=['populationTotal', 'GNIperCap','domain','type','group'])
table_1 = table_1.dropna(subset=['value'])
agg = table_1.groupby('indicator', as_index=False)['date'].agg(['min', 'max']).reset_index()
dates = agg.melt(id_vars='indicator', value_vars=['min', 'max'], value_name='date')[['indicator', 'date']].drop_duplicates()
table_1 = table_1.merge(dates, on=['indicator', 'date'], how='inner').sort_values(['indicator', 'date']).reset_index(drop=True)
table_1 = table_1.sort_values(by=['dimension','indicator']).reset_index(drop=True)
table_1['value']= round(table_1['value'],2)
table_1 = pivot_table(
    table_1,
    index=['dimension', 'indicator'],
    values=['date', 'value'],
    aggfunc=['first', 'last']
).reset_index()
new_cols = []
for col in table_1.columns:
    if isinstance(col, tuple):
        new_cols.append((col[1], col[0]))
    else:
        new_cols.append((col, ''))
table_1.columns = pd.MultiIndex.from_tuples(new_cols)
no_change = table_1[('value', 'first')] == table_1[('value', 'last')]
table_1.loc[no_change, ('date', 'last')] = table_1.loc[no_change, ('date', 'first')]
table_1 = table_1.sort_index(axis=1, level=0)
table_1[('','indicator')] = [
    "Population not accessing the internet",
    "Urban population lacking convenient access to public transport",
    "Adult population (aged 15+ years) who are illiterate",
    "Young adult population (aged 21-23 years) with incomplete upper secondary education",
    "Population lacking access to electricity",
    "Population lacking access to clean fuels and technologies for cooking, heating and lighting",
    "Population-weighted score on the Gender Inequality Index (global gap between women and men in terms of reproductive health, empowerment and employment)",
    "Population with moderate to severe food insecurity",
    "Population undernourished",
    "Population living in countries without high coverage of essential health services (Universal Health Coverage Index score less than 60 out of 100)",
    "Population living in countries with under-5 mortality rate exceeding 25 per 1,000 live births",
    "Urban population living in slums or informal settlements",
    "Population living below the societal poverty line, set at half their country's median household income or at least $15 a day",
    "Population of young people (aged 15-24 years) not in employment, education or training",
    "Population stating that they perceive widespread corruption in government and business",
    "Population living in countries with a homicide rate of 5 or more per 100,000",
    "Population living in countries governed by an autocratic regime",
    "Population living in countries with a Palma ratio of 2 or more (the income share of the richest 10% of people relative",
    "Population stating that they are without someone to count on in times of trouble",
    "Population lacking access to safely managed drinking water",
    "Population lacking access to safely managed sanitation",
]
styled_html_1 = table_1.to_html(index=False)
styled_html_1 = styled_html_1.replace('<table', '<table style="text-align: left"')
styled_html_1 = styled_html_1.replace('<th', '<th style="text-align: left; font-weight: bold"')
styled_html_1 = styled_html_1.replace('<td', '<td style="text-align: left"')
display(HTML(styled_html_1))
```

# Ecological overshoot clustering

Continuing to the plots, we see some barplots referred to the ecological overshoot. Now I used an already split into country categories dataset. The approach of splitting is based on the GNI per capita. The split creates 3 groups: Bottom 40, Middle 40 and Top 20. The visual shows the impact of each group to each of the indicators.

### Country-group comparison: social indicators

This section applies the same country-group perspective to social indicators. Values describe each indicator using its defined unit and should not be compared as if unlike measures shared one scale.

```python
import matplotlib.pyplot as plt
from matplotlib.ticker import PercentFormatter
df_1 = pd.read_csv('11_20250115_multiPanelData.csv')  
trellis = df_1[(df_1['domain'] == 'ecological') & (df_1['type'] == 'national aggregate') & (df_1['group']!='World')].reset_index(drop=True)
trellis = trellis.dropna(subset=['ratio'])
trellis = trellis.drop(columns=['populationTotal', 'dimension','domain','type','grpCode', 'indCode','date','GNIperCap','value','boundary'])
trellis ['ratio'] = round(trellis ['ratio'] * 100,0)
colors = {'Bottom-40': '#FFD700', 'Middle-40': '#DDA0DD', 'Top-20': '#4682B4'}
fig, axes = plt.subplots(2, 3, figsize=(10, 8))
axes = axes.flatten()
trellis['indicator'] = trellis['indicator'].astype('string')
trellis['indicator'] = trellis['indicator'].replace({
    "blueH2O_footprint": "Blue water footprint",
    "co2_footprint": "Carbon footprint",
    "hanpp_footprint": "HANPP footprint",
    "N_footprint": "Nitrogen footprint",
    "P_footprint": "Phosphorus footprint",
    "species_footprint": "Species-loss footprint",
})
trellis = trellis.sort_values("indicator", key=lambda s: s.str.lower())
indicators = trellis["indicator"].unique()
for idx, indicator in enumerate(indicators):
    data = trellis[trellis["indicator"] == indicator].groupby("group")["ratio"].mean()
    bar_colors = [colors.get(group, "steelblue") for group in data.index]
    axes[idx].bar(data.index, data.values, color=bar_colors)
    axes[idx].set_title(indicator)
    axes[idx].set_ylim(0, 1000)
    axes[idx].yaxis.set_major_formatter(PercentFormatter())
    axes[idx].tick_params(axis="x", rotation=45, labelbottom=False, labeltop=False)
    if idx < 3:
        axes[idx].tick_params(axis="x", labeltop=True)
        axes[idx].xaxis.set_ticks_position("top")

for idx in range(len(indicators), len(axes)):
    axes[idx].axis("off")
ncols = 3
nrows = 2
for r in range(nrows):
    left_ax = axes[r * ncols]
    left_ax.set_ylabel("Ecological overshoot",fontweight='bold')    
plt.tight_layout()
plt.show()
```

# Sandwich Plot

As the title describes, we have 2 barplots that expand to exactly opposite direction starting from the center. There are not any country groups, so the plot is referred to the world's results. In each barplot the normal red color describes the ratio that each indicator has on the beginning of being recorded. The light red color represents the reduction from ending value to the starting one. On the contrary, the dark red color represents the increase from the starting value to the ending one. Upper baplot has a scale of 0 - 300 while the lower one has 0 - 100. Species extinctions and hazardus chemical production are way more than the limits. That means even if we have an improvement to these 2 factors it is not enough.

- the sum of the dark‑red and regular‑red portions of the bar (if the indicator was worsening), or
- only the regular‑red portion (if the indicator was improving).

The color intensity reflects the deviation from the value of 100%:
lighter shades indicate larger deviations, while darker shades indicate smaller ones.

The indicators are also grouped by dimension. Immediately outside the inner ring (dark green), which labels the individual indicators, there is a lighter green ring that labels the broader dimensions to which these indicators belong.

In contrast, the outer ring refers to the ecological indicators. Here, the slices expand outward, and the grouping logic is similar: indicators are organized by dimension, and each dimension is displayed directly above its corresponding slices.

Finally, the plot is fully interactive. Hovering over any slice reveals the indicator’s name along with its corresponding ratio value.
 

```python
doughnut = pd.read_csv("5_20250108_doughnutDat-boundariesRatios.csv")
doughnut = doughnut[(doughnut["type"] == "global doughnut")
                    & (doughnut["date"] == 2022)
                    & (doughnut["indicator"] != "racialInequality")].reset_index(drop=True)
interpretation = pd.read_csv("12_20250515_globalDoughnutData_2000-2022.csv")
interpretation = interpretation [(interpretation['type'] == "global doughnut")
                    & (interpretation["date"] == 2022)
                    & (interpretation["indicator"] != "racialInequality")].reset_index(drop=True)
doughnut = doughnut.merge(
    interpretation[["indCode", "unit", "shortfallOvershoot_pct"]],
    on=["indCode"],
    how="left",
)
doughnut.loc[doughnut["domain"] == "ecological", "ratio"] = doughnut.loc[
    doughnut["domain"] == "ecological", "ratio"
].sub(1)
doughnut["ratio"] = np.where(doughnut["ratio"] < 0, 0, doughnut["ratio"])
doughnut["ratio"] = (doughnut["ratio"] * 100).round(2)
doughnut["indicator"] = doughnut["indicator"].replace({
    "undernourishment": "Undernourished",
    "foodInsecurity": "Food insecurity",
    "under5death": "Child mortality",
    "UHCindex": "Lack of health services",
    "adultLiteracy": "Illiteracy",
    "secondarySchool": "Incomplete secondary school",
    "societalPoverty": "Societal poverty",
    "youthNEET": "Youth unemployment",
    "drinkingH2O": "Unsafe drinking water",
    "sanitation": "Unsafe sanitation",
    "energyAccess": "Lack of electricity",
    "energyIndoor": "Lack of clean fuels indoors",
    "publicTrans": "Lack of public transport",
    "internet": "Lack of internet",
    "urbanSlums": "Inadequate housing",
    "genderGapIndex": "Gender inequality",
    "socialSupport": "Lack of social support",
    "palma": "Income inequality",
    "govRegimes": "Autocratic regimes",
    "controlCorruption": "Perceptions of corruption",
    "homicideOver5": "Homicides",
    "blueDev": "Blue-water flows",
    "co2_ppm": "CO2 concentration",
    "hanppGtC": "Human appropriation of energy production",
    "nitrogenMt": "Nitrogen polution",
    "phosphorusMt": "Phosphorus polution",
    "extinction1900": "Species extinctions",
    "totalOzone": "Stratospheric ozone concentration",
    "forestAreaMKM2": "Forest area",
    "interhemAOD": "Aerosol optical depth",
    "chemicalsMt_Hzd": "Hazardous chemicals production",
    "soilDev": "Soil moisture",
    "omega_a": "Aragonite saturation",
    "erf_wm2": "Radioactive forcing",
})
eco = doughnut[doughnut["domain"] == "ecological"].copy()
soc = doughnut[doughnut["domain"] == "social"].copy()
eco = eco.sort_values(["dimension","indicator"]).reset_index(drop=True)
soc = soc.sort_values(["dimension","indicator"]).reset_index(drop=True)
soc['ratio'] = 100 - soc['ratio']
import plotly.graph_objects as go
from matplotlib.colors import to_hex
from matplotlib.colors import LinearSegmentedColormap
def color_scale_fixed(values, color_start, color_end):
    min_val = values.min()
    max_val = values.max()
    if max_val == min_val:
        normalized = [0.5] * len(values)
    else:
        normalized = (values - min_val) / (max_val - min_val)
    cmap = LinearSegmentedColormap.from_list("gradient", [color_start, color_end])
    return [cmap(val) for val in normalized]
soc_colors_hex = [to_hex(c) for c in color_scale_fixed(soc["ratio"], "#E88AA0", "#8B1E5B")]
eco_colors_hex = [to_hex(c) for c in color_scale_fixed(eco["ratio"], "#E88AA0", "#8B1E5B")]
ring_inner = 1.0
ring_width = 0.2
ring_outer = ring_inner + ring_width
ring_light_width = 0.4
ring_light_outer = ring_outer + ring_light_width
ring_dark_width = 0.2
ring_dark_outer = ring_light_outer + ring_dark_width
n_soc = len(soc)
n_eco = len(eco)
soc_theta = np.linspace(0, 2 * np.pi, n_soc, endpoint=False)
eco_theta = np.linspace(0, 2 * np.pi, n_eco, endpoint=False)
soc_width = 2 * np.pi / n_soc
eco_width = 2 * np.pi / n_eco
soc_heights = soc["ratio"] / 100
eco_heights = (eco["ratio"] / 100) / 10
fig = go.Figure()
for i in range(n_soc):
    theta_range = np.linspace(soc_theta[i], soc_theta[i] + soc_width, 30)
    r_outer = ring_inner
    x_outer = r_outer * np.cos(theta_range)
    y_outer = r_outer * np.sin(theta_range)
    r_inner = ring_inner - soc_heights.iloc[i]
    x_inner = r_inner * np.cos(theta_range)
    y_inner = r_inner * np.sin(theta_range)
    x_poly = np.concatenate([x_outer, x_inner[::-1], [x_outer[0]]])
    y_poly = np.concatenate([y_outer, y_inner[::-1], [y_outer[0]]])
    indicator_name = soc["indicator"].iloc[i]
    ratio_val = soc["ratio"].iloc[i]
    fig.add_trace(go.Scatter(
        x=x_poly,
        y=y_poly,
        fill="toself",
        fillcolor=soc_colors_hex[i],
        line=dict(color="white", width=1),
        text=[f"{indicator_name}<br>Ratio: {ratio_val:.2f}%"] * len(x_poly),
        hoverinfo="text",
        hoveron="fills+points",
        mode="lines+markers",
        marker=dict(size=2, opacity=0),
        showlegend=False,
        name=""
    ))
gap_ring_width = -0.5
inner_gap_outer = ring_inner
inner_gap_inner = ring_inner - gap_ring_width
theta_circle = np.linspace(0, 2 * np.pi, 240)
x_outer = inner_gap_outer * np.cos(theta_circle)
y_outer = inner_gap_outer * np.sin(theta_circle)
x_inner = inner_gap_inner * np.cos(theta_circle)
y_inner = inner_gap_inner * np.sin(theta_circle)
fig.add_trace(go.Scatter(
    x=np.concatenate([x_outer, x_inner[::-1], [x_outer[0]]]),
    y=np.concatenate([y_outer, y_inner[::-1], [y_outer[0]]]),
    fill="toself",
    fillcolor="#227444",
    line=dict(color="#227444", width=0),
    hoverinfo="skip",
    showlegend=False,
    mode="lines"
))
light_ring_inner = ring_outer
light_ring_outer = ring_light_outer
theta_circle = np.linspace(0, 2 * np.pi, 240)
x_outer = light_ring_outer * np.cos(theta_circle)
y_outer = light_ring_outer * np.sin(theta_circle)
x_inner = light_ring_inner * np.cos(theta_circle)
y_inner = light_ring_inner * np.sin(theta_circle)
fig.add_trace(go.Scatter(
    x=np.concatenate([x_outer, x_inner[::-1], [x_outer[0]]]),
    y=np.concatenate([y_outer, y_inner[::-1], [y_outer[0]]]),
    fill="toself",
    fillcolor="#6fb447",
    line=dict(color="#6fb447", width=0),
    hoverinfo="skip",
    showlegend=False,
    mode="lines"
))
outer_ring_inner = ring_light_outer
outer_ring_outer = ring_dark_outer
theta_circle = np.linspace(0, 2 * np.pi, 240)
x_outer = outer_ring_outer * np.cos(theta_circle)
y_outer = outer_ring_outer * np.sin(theta_circle)
x_inner = outer_ring_inner * np.cos(theta_circle)
y_inner = outer_ring_inner * np.sin(theta_circle)
fig.add_trace(go.Scatter(
    x=np.concatenate([x_outer, x_inner[::-1], [x_outer[0]]]),
    y=np.concatenate([y_outer, y_inner[::-1], [y_outer[0]]]),
    fill="toself",
    fillcolor="#227444",
    line=dict(color="#227444", width=0),
    hoverinfo="skip",
    showlegend=False,
    mode="lines"
))
soc_dims = soc.groupby("dimension", sort=False, group_keys=False)[["indicator"]].apply(
    lambda g: (g.index[0], g.index[-1]), include_groups=False
).to_dict()
label_radius_soc = (ring_outer + ring_light_outer) / 2
for dim_name, (start_idx, end_idx) in soc_dims.items():
    mid_theta = soc_theta[start_idx] + (soc_width * (end_idx - start_idx + 1) / 2)
    label_x = label_radius_soc * np.cos(mid_theta)
    label_y = label_radius_soc * np.sin(mid_theta)
    rotation = 90 - np.degrees(mid_theta)
    formatted_text = dim_name.replace(" ", "<br>")

    fig.add_annotation(
        x=label_x,
        y=label_y,
        text=formatted_text,
        showarrow=False,
        textangle=rotation,
        font=dict(size=10, color="white", family="Arial, sans-serif", weight="bold")
    )
text_radius = ring_inner - 0.08
text_string = "SOCIAL FOUNDATION"
char_count = len(text_string)
total_arc = np.pi * 0.5
char_angles = np.linspace(np.pi / 2 + total_arc / 2, np.pi / 2 - total_arc / 2, char_count)
for i, char in enumerate(text_string):
    char_theta = char_angles[i]
    char_x = text_radius * np.cos(char_theta)
    char_y = text_radius * np.sin(char_theta) + 0.18
    rotation_angle = 90 - np.degrees(char_theta)
    fig.add_annotation(
        x=char_x,
        y=char_y,
        text=char,
        showarrow=False,
        textangle=rotation_angle,
         font=dict(
            size=12,
            color="white",
            family="Arial, sans-serif",
            weight="bold"
        ))
text_radius_1 = ring_dark_outer + 0.08
text_string_1 = "ECOLOGICAL CEILING"
char_count_1 = len(text_string_1)
total_arc_1 = np.pi * 0.4
char_angles_1 = np.linspace(np.pi / 2 + total_arc_1 / 2, np.pi / 2 - total_arc_1 / 2, char_count_1)
for j, char_1 in enumerate(text_string_1):
    char_theta_1 = char_angles_1[j]
    char_x_1 = text_radius_1 * np.cos(char_theta_1)
    char_y_1 = text_radius_1 * np.sin(char_theta_1) -0.22
    rotation_angle_1 = 90 - np.degrees(char_theta_1)
    fig.add_annotation(
        x=char_x_1,
        y=char_y_1,
        text=char_1,
        showarrow=False,
        textangle=rotation_angle_1,
         font=dict(
            size=12,
            color="white",
            family="Arial, sans-serif",
            weight="bold"
        ))
eco_dims = eco.groupby("dimension", sort=False, group_keys=False)[["indicator"]].apply(
 lambda g: (g.index[0], g.index[-1]), include_groups=False).to_dict()
label_radius_eco = ring_dark_outer + (eco_heights.max() *0.10)
for dim_name, (start_idx, end_idx) in eco_dims.items():
 mid_theta = eco_theta[start_idx] + (eco_width * (end_idx - start_idx +1) /2)
 label_x = label_radius_eco * np.cos(mid_theta)
 label_y = label_radius_eco * np.sin(mid_theta)
 rotation =90 - np.degrees(mid_theta)
 formatted_text = dim_name.replace(" ", "<br>")
 fig.add_annotation(
 x=label_x,
 y=label_y,
 text=formatted_text,
 showarrow=False,
 textangle=rotation,
 font=dict(size=10, color="black", family="Arial, sans-serif", weight="bold")
 )
for i in range(n_eco):
    theta_range = np.linspace(eco_theta[i], eco_theta[i] + eco_width, 30)
    r_inner = ring_dark_outer
    x_inner = r_inner * np.cos(theta_range)
    y_inner = r_inner * np.sin(theta_range)
    r_outer = ring_dark_outer + eco_heights.iloc[i]
    x_outer = r_outer * np.cos(theta_range)
    y_outer = r_outer * np.sin(theta_range)
    x_poly = np.concatenate([x_inner, x_outer[::-1], [x_inner[0]]])
    y_poly = np.concatenate([y_inner, y_outer[::-1], [y_inner[0]]])
    indicator_name = eco["indicator"].iloc[i]
    ratio_val = eco["ratio"].iloc[i]
    fig.add_trace(go.Scatter(
        x=x_poly,
        y=y_poly,
        fill="toself",
        fillcolor=eco_colors_hex[i],
        line=dict(color="white", width=1),
        text=[f"{indicator_name}<br>Ratio: {ratio_val:.2f}%"] * len(x_poly),
        hoverinfo="text",
        hoveron="fills+points",
        mode="lines+markers",
        marker=dict(size=2, opacity=0),
        showlegend=False,
        name=""
    ))

fig.update_layout(
    xaxis=dict(showgrid=False, zeroline=False, showticklabels=False, scaleanchor="y", scaleratio=1),
    yaxis=dict(showgrid=False, zeroline=False, showticklabels=False, scaleanchor="x", scaleratio=1),
    hovermode="closest",
    plot_bgcolor="white",
    paper_bgcolor="white",
    width=900,
    height=900,
    margin=dict(l=50, r=50, t=50, b=50),
)
fig.show()
```
